```python
Deep and Shallow Copy
```

```python
b = np.arange(1,31)
b
```

```python
a = b
a
```

```python
a[0] = 23
a
```

```python
# Here the Deep copy working
# Ram is consisting of a and b array data at the same location in ram
# That's why both are changing if one is changed
b
```

```python
# Shallow 
a = b.copy()
a
```

```python
# Here shallow copy is working where the data of the a is changed whereas the data in the b is remain unchanged.
a[0] =34
b
```
