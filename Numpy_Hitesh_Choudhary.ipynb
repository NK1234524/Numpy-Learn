```python
# Sorting
unsorted = np.array([1,4,2,7,5,8,3])
unsorted.sort() # Simplest way to sort an array
unsorted
```

```python
arr_2D_unsorted = np.array([[3,1],[1,2],[2,3]])
new_arr = np.sort(arr_2D_unsorted, axis=0) # column sort
new_arr = np.sort(arr_2D_unsorted, axis=1) # row sort
new_arr
```

```python
# filter
numbers = np.array([1,2,3,5,6,7,8])
even_numbers = numbers[numbers%2==0]
even_numbers
```

```python
# filter with mask
# mask is the term which is used to evaluate the value.
mask  = numbers>5
print(numbers[mask])
```

```python
# fancy indexing vs np.where()
indices =  [0,2,4]
numbers[indices] # so similar to the mask
where_result = np.where(numbers>5)
numbers[where_result]
```

```python
# np.where best application
result = np.where(numbers>5 , numbers+1 , numbers-1) #(condition,True,False)
result
```

```python
# Adding and removing data
# Combining the arrays
arr1 = np.arange(1,21)
arr2 = np.arange(21,42)
combined = np.concatenate((arr1,arr2))
combined
```

```python
# Array Compatibility
arr1 = np.arange(1,11)
arr2 = np.arange(11,21)
arr3 = np.arange(21 , 31)
print("Compatibilty :" , arr1.shape == arr2.shape ) # Compatibility  '==' this operator is used here 
```

```python
# Multidimensional array operations
arr1 = np.array([[1,3],[5,6],[9,4]])
arr2 = np.array([[10,11],[7,2],[1,3]])



arr_row = np.array([[3,2],[6,7]])
arr_col = np.array([[5],[1],[2]])
# To add new row and column we will use vstack and hstack
# Dimensions should also match here
new_arr_here = np.vstack((arr1,arr_row))


new_arr_col = np.hstack((arr2,arr_col))
# new_arr_col
new_arr_here
```
