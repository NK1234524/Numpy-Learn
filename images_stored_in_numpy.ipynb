```python
import numpy as np
import matplotlib.pyplot as plt

logo = np.random.rand(100,100)
np.save('numpy-logo.npy',logo)
```

```python
array1 = np.arange(1,10)
array2 = np.random.rand(3,3)
array3 = np.zeros((4,4))
array4 = np.array([[1,2],[3,4],[5,6]])
```

```python
# In Numpy np.save is used to save the matrix value in the file such that the image generated from the 
# prediction of the generative model is then stored in the .npy file 
# Without .npy the data is just stored in the form of Ram run time memory 

np.save('array.npy',array1)
np.save('array2.npy',array2)
np.save('array3.npy',array3)
```

```python
# Now after saving the data of the matrix now we also need to store the info in the form of matrix
# Now we need to load the file here
loaded_array= np.load('array.npy')
loaded_array1 = np.load('array2.npy')
print(loaded_array1)
print(loaded_array)
```
