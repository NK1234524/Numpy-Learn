```python
data = np.array([
    [18, 85, 78],   # Student 1
    [19, 92, 88],   # Student 2
    [17, 76, 95],   # Student 3
    [18, 65, 70],   # Student 4
    [20, 90, 85]    # Student 5
])
```

```python

# Get the shape of the matrix.
sh = data.shape # Shape is an instance attribute so we don't need to put the () here.
sh
```

```python
# Find the average age of students.
ans =np.mean(data[0:5,0])
ans
```

```python
#Extract Math marks of all students.
mm = data[1,0:5]
mm
```

```python
#Find the highest Science mark.
ss = np.max(data[2,0:5])
ss
```

```python
#Get details of the student who scored more than 90 in Math.
values = data[data[:,1]>90]
values    # indexing concept
```

```python
#Increase Math marks of all students by 5.
data[:,1] +=5 # Vectorization & broadcasting
data
```

```python
#Find how many students are younger than 19.
age =data[data[:,0]<19]
age
```

```python
#Calculate the average marks in each subject (column-wise mean).
gh = data[:,1].mean()
gh1 = data[:,2].mean()
print(f"mean of maths :{gh}")
print(f"mean of Sceince : {gh1}")
```

```python
#Get data of students who scored at least 80 in both subjects.
marks = data[:,1:3]<=80 # vectorization & broadcasting
marks
```
